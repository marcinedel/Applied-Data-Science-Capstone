## Task 1: Mark all launch sites on a map


Najpierw spróbujmy nanieść lokalizację każdego obiektu na mapę, korzystając z jego współrzędnych geograficznych – szerokości i długości geograficznej.

Poniższy zbiór danych, o nazwie `spacex_launch_geo.csv`, jest zbiorem rozszerzonym – dla każdej lokalizacji dodano w nim współrzędne geograficzne (szerokość i długość geograficzną).

```python
# Download and read the `spacex_launch_geo.csv`
from js import fetch
import io

URL = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/spacex_launch_geo.csv'
resp = await fetch(URL)
spacex_csv_file = io.BytesIO((await resp.arrayBuffer()).to_py())
spacex_df=pd.read_csv(spacex_csv_file)
```

Teraz możesz sprawdzić, jakie są współrzędne dla poszczególnych lokalizacji.

```python
# Select relevant sub-columns: `Launch Site`, `Lat(Latitude)`, `Long(Longitude)`, `class`
spacex_df = spacex_df[['Launch Site', 'Lat', 'Long', 'class']]
launch_sites_df = spacex_df.groupby(['Launch Site'], as_index=False).first()
launch_sites_df = launch_sites_df[['Launch Site', 'Lat', 'Long']]
launch_sites_df
```

Najpierw musimy utworzyć obiekt `Map` biblioteki folium, ustawiając początkowy punkt centralny na Centrum Kosmiczne im. Johnsona (NASA Johnson Space Center) w Houston w stanie Teksas.

```python
# Start location is NASA Johnson Space Center
nasa_coordinate = [29.559684888503615, -95.0830971930759]
site_map = folium.Map(location=nasa_coordinate, zoom_start=10)
```

Możemy użyć `folium.Circle`, aby dodać wyróżniony obszar w kształcie koła wraz z etykietą tekstową w określonym miejscu (według współrzędnych). Na przykład:

```python
# Create a blue circle at NASA Johnson Space Center's coordinate with a popup label showing its name
circle = folium.Circle(nasa_coordinate, radius=1000, color='#d35400', fill=True).add_child(folium.Popup('NASA Johnson Space Center'))
# Create a blue circle at NASA Johnson Space Center's coordinate with a icon showing its name
marker = folium.map.Marker(
    nasa_coordinate,
    # Create an icon as a text label
    icon=DivIcon(
        icon_size=(20,20),
        icon_anchor=(0,0),
        html='<div style="font-size: 12; color:#d35400;"><b>%s</b></div>' % 'NASA JSC',
        )
    )
site_map.add_child(circle)
site_map.add_child(marker)
```

# Task 2: Mark the success/failed launches for each site on the map


Następnie spróbujmy wzbogacić mapę, dodając informacje o wynikach startów dla każdej lokalizacji, aby sprawdzić, w których z nich odnotowano wysoki odsetek sukcesów.
Przypomnijmy, że ramka danych `spacex_df` zawiera szczegółowe rejestry startów, a kolumna `class` wskazuje, czy dany start zakończył się powodzeniem, czy nie.

```python
spacex_df.tail(10)
```

Następnie utwórzmy znaczniki dla wszystkich rekordów dotyczących startów.
Jeśli start zakończył się powodzeniem `(class=1)`, użyjemy zielonego znacznika, a jeśli zakończył się niepowodzeniem – czerwonego znacznika `(class=0)`.

Warto zauważyć, że starty odbywają się tylko w jednym z czterech wyznaczonych miejsc, co oznacza, że ​​wiele rekordów dotyczących startów będzie miało dokładnie te same współrzędne. Grupowanie znaczników (ang. *marker clusters*) może być dobrym sposobem na uproszczenie mapy zawierającej wiele znaczników o tych samych współrzędnych.

Najpierw utwórzmy obiekt `MarkerCluster`

```python
marker_cluster = MarkerCluster()
```

*TODO:* Utwórz nową kolumnę o nazwie `marker_color` w ramce danych `spacex_df`, w której będą przechowywane kolory znaczników w zależności od wartości w kolumnie `class`

```python
# Apply a function to check the value of `class` column
def assign_marker_color(launch_outcome):
    # If class=1, marker_color value will be green
    if launch_outcome == 1:
        return 'green'
    # If class=0, marker_color value will be red
    else:
        return 'red'
# Funkcja przypisująca kolor do wyniku startu    
spacex_df['marker_color'] = spacex_df['class'].apply(assign_marker_color)
spacex_df.tail(10)
```

*TODO:* Dla każdego wyniku startu w ramce danych `spacex_df` dodaj obiekt `folium.Marker` do `marker_cluster`

```python
# Add marker_cluster to current site_map
site_map.add_child(marker_cluster)

# for each row in spacex_df data frame
# create a Marker object with its coordinate
# and customize the Marker's icon property to indicate if this launch was successed or failed, 
# e.g., icon=folium.Icon(color='white', icon_color=row['marker_color']
for index, record in spacex_df.iterrows():
    marker = folium.Marker([record['Lat'], record['Long']], 
                  icon=folium.Icon(color='white', icon_color=record['marker_color']))
    marker_cluster.add_child(marker)

site_map
```

# TASK 3: Calculate the distances between a launch site to its proximities


Następnie musimy zbadać i przeanalizować bliskość miejsc startu.

Najpierw dodajmy na mapie element `MousePosition`, dzięki któremu będzie można odczytać współrzędne punktu, nad którym znajduje się kursor myszy. Dzięki temu, podczas przeglądania mapy, łatwo sprawdzisz współrzędne dowolnych interesujących Cię miejsc (na przykład linii kolejowych).

```python
# Add Mouse Position to get the coordinate (Lat, Long) for a mouse over on the map
formatter = "function(num) {return L.Util.formatNum(num, 5);};"
mouse_position = MousePosition(
    position='topright',
    separator=' Long: ',
    empty_string='NaN',
    lng_first=False,
    num_digits=20,
    prefix='Lat:',
    lat_formatter=formatter,
    lng_formatter=formatter,
)

site_map.add_child(mouse_position)
site_map
```

Teraz przybliż widok na miejsce startu i zbadaj jego otoczenie, sprawdzając, czy w pobliżu znajdują się linie kolejowe, drogi, linia brzegowa itp. Wskaż kursorem te obiekty i zanotuj ich współrzędne (widoczne w lewym górnym rogu), aby określić odległość od miejsca startu.

Odległość między dwoma punktami na mapie można obliczyć na podstawie ich współrzędnych geograficznych (szerokości i długości geograficznej), stosując następującą metodę:

```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```
